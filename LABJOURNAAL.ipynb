### Labjournaal double pendulum

Begonnen met lezen van de bestanden op brightspace, duidelijk krijgen wat de bedoeling is. code openen opencv downloaden, juiste librarys. Vervolgens de pendulum meten que lengte en massa van de twee pendulums. Punten die we willen tracken markeren met stift blauw punt op draaipunt, rood punt op uiteinde. Kalibratiemeting, foto van de pendulum met een liniaal erbij. Punten zijn te klein, groter gemaakt( groen en rood). Video maken, en proberen in te laden via draadje op laptop gezet. Dit is gelukt, alleen nu trackt onze code alleen het rode punt. Hierna proberen de code aan te passen door de cel van de rode marker te kopieren, de hue aan te passen zodat ie groen trackt, en later in de tracking ook de groene code aanhalen. Plotje gemaakt, het valt op dat er een groot aantal datapunten linksboven in de hoek vallen, waarschijnlijk doordat er linksboven ook een rood iets was wat de code trackt ipv het rode punt op onze pendulum.
